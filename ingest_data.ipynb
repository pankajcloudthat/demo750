### Ingest SQL Data

```python
# Read data from MySQL table 'customer' and insert into Databricks table 'samples.tpch.customer'
mysql_url = "jdbc:mysql://<mysql-host>:<mysql-port>/<database-name>"
mysql_properties = {
    "user": "<mysql-username>",
    "password": "<mysql-password>",
    "driver": "com.mysql.cj.jdbc.Driver"
}

# Load MySQL data into a Spark DataFrame
customer_df = spark.read.jdbc(
    url=mysql_url,
    table="customer",
    properties=mysql_properties
)

# Write the DataFrame into Databricks table
customer_df.write.mode("append").saveAsTable("samples.tpch.customer")
```
